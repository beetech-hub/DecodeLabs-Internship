## **PHASE 2 — Missing Value Analysis & Handling**

### #2.1 Check Missing Values

```python
df.isnull().sum()
```

### #2.2 Analyze Missing Values

#### Findings

- A missing value analysis was performed to identify incomplete records.

- The results show that all columns are complete except for the `CouponCode` column, which contains 309 missing values.

- This is likely due to transactions where customers did not apply
a promotional coupon during checkout.

### #2.3 Handle Missing Values

```python
df['CouponCode'] = df['CouponCode'].fillna('No Coupon')
```

### #2.4 Re-Check Missing Values

```python
df.isnull().sum()
```

## **PHASE 3 — Duplicate Detection & Removal**

### #3.1 Check Duplicate Records

```python
df.duplicated().sum()
```

Analysis of Duplicate Data

- A duplicate check was performed using
the `duplicated().sum()` function.

- The result shows that no duplicate records
were found in the dataset.

### #3.2 Remove Duplicate Records

```python
df.drop_duplicates(inplace=True)
```

### #3.3 Validate Duplicate Removal

```python
df.duplicated().sum()
```

- The dataset was revalidated after the duplicate handling process.

- The validation confirms that the dataset
contains zero duplicate records.

## **PHASE 4 — Data Formatting, Cleaning, & Validation**

### #4.1 Check Data Types

```python
df.info()
```

### #4.2 Date Formatting

```python
df['Date'] = pd.to_datetime(df['Date'])
```

The `Date` column was converted into
datetime format to ensure consistency
and compatibility for future analysis.

### #4.3 Text Cleaning

```python
text_columns = ['Product', 'ShippingAddress', 'PaymentMethod',
                'OrderStatus', 'ReferralSource']

for col in text_columns:
    df[col] = df[col].str.strip()
```

Text-based columns were cleaned using
the `str.strip()` method to remove
unnecessary leading and trailing spaces.

### #4.4 Numeric Data Validation

```python
df[['Quantity', 'UnitPrice', 'ItemsInCart', 'TotalPrice']].dtypes
```

Numeric columns were checked to ensure
they use appropriate numerical data types
for accurate calculations and analysis.
