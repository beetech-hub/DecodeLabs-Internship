###  #6. DISTRIBUTION ANALYSIS

```python
plt.figure(figsize=(8,5))

sns.histplot(df['TotalPrice'], kde=True)

plt.title('Sales Distribution')
plt.xlabel('TotalPrice')
plt.ylabel('Frequency')

plt.show()
```

#### # Purpose:
- Understand data distribution
- Detect skewness
- Observe concentration of values

#### # Result:

1. The TotalPrice distribution is positively skewed.
2. Most transactions are concentrated in the low-to-medium price range.
3. A small number of transactions contain exceptionally high total prices.
4. High-value transactions significantly influence the overall average sales value.

### #7. MEAN VS MEDIAN

```python
mean_sales = df['TotalPrice'].mean()
median_sales = df['TotalPrice'].median()

print("Mean :", mean_sales)
print("Median :", median_sales)
```

#### # Interpretation:
- Mean > Median → Right-Skewed Distribution
- Mean < Median → Left-Skewed Distribution
- Mean ≈ Median → Symmetrical Distribution

#### #Result:
The mean TotalPrice is significantly higher than the median, indicating that several high-value transactions pull the average upward.

### #9. BOXPLOT ANALYSIS

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df['TotalPrice'])

plt.title('Boxplot of TotalPrice')

plt.show()
```

#### Purpose:
- Detect outliers
- Understand spread of data
- Visualize distribution

#### #Result:

1. The boxplot reveals several high-value outliers beyond the upper whisker.
2. Most transactions fall within a moderate TotalPrice range.
3. The presence of outliers indicates that a small number of transactions differ significantly from the majority of observations.
4. These outliers may represent bulk purchases or premium product orders.

### #10. OUTLIER DETECTION USING IQR

```python
Q1 = df['TotalPrice'].quantile(0.25)
Q3 = df['TotalPrice'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['TotalPrice'] < lower_bound) |
    (df['TotalPrice'] > upper_bound)
]

print("Jumlah Outlier:", outliers.shape[0])

outliers.head()
```

### #11. CORRELATION ANALYSIS

```python
corr = df.corr(numeric_only=True)

corr
```

#### # Purpose:
- Identify relationships between variables
- Detect positive/negative correlation

### #12. CORRELATION HEATMAP

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```

#### #Insight:
1. UnitPrice ↔ TotalPrice (0.72)
- UnitPrice shows a strong positive correlation with TotalPrice, indicating that product price significantly influences total transaction value.
2. Quantity ↔ TotalPrice (0.62)
- Quantity has a strong positive relationship with TotalPrice, meaning larger purchase quantities contribute to higher transaction values.
3. Quantity ↔ ItemsInCart (0.65)
- Quantity and ItemsInCart show a strong positive correlation, suggesting that customers purchasing more items tend to buy larger quantities overall.
4. ItemsInCart ↔ TotalPrice (0.39)
- ItemsInCart shows only a moderate correlation with TotalPrice, indicating that transaction value depends not only on item count but also on product pricing.
